# YojanaMitraAI – Backend Contribution Documentation

## AI-Based Government Scheme Recommendation System

Team: Backend Development Team  
Technology: FastAPI + Python


# Project Overview

YojanaMitraAI is an AI-powered government scheme recommendation system designed to help users discover suitable welfare schemes based on eligibility criteria such as:

- Age
- Gender
- Income
- Occupation
- Education
- State
- Category

The system uses:
- FastAPI backend services
- NLP-based recommendation pipeline
- Rule-based eligibility filtering
- Semantic search and ranking

# Backend Architecture

The backend of YojanaMitraAI is developed using FastAPI.

Main responsibilities of backend:
- Handle API requests
- Process user eligibility data
- Connect with recommendation engine
- Retrieve schemes from database
- Return filtered recommendations
- Manage notifications and authentication

Backend Stack:
- FastAPI
- Python
- REST APIs
- JSON Serialization
- Supabase Integration

# API Workflow

1. User submits eligibility details from frontend
2. Request sent to FastAPI backend
3. Backend validates incoming data
4. Recommendation pipeline processes request
5. Eligibility filtering is applied
6. Matching schemes are ranked
7. Response returned to frontend dashboard

# Example API Endpoint Structure

Example Endpoint:
POST /recommend-schemes

Purpose:
Returns eligible government schemes based on user profile.

Input:
- Age
- Gender
- State
- Occupation
- Income

Output:
- Recommended schemes
- Eligibility score
- Missing criteria
